```python
List1=["apple","banana","cherry"]
print("List Elements are", List1)
print("Access List element:", List1[2])
List1.append("mango")
print ("Add List Item:", List1)
List1.remove("apple")
print ("Remove List Item:", List1)
List1.sort()
print ("Sort List Item:", List1)
List2=["Berry","Watermelon","Guava"]
List3=List1+List2
print ("Combine Two Lists:", List3)
```

```python
Tuple1=("Red","Orange","Yellow","Blue")
print("Tuple Elements are:", Tuple1)
print("Access Tuple element:", Tuple1 [2])
p= list(Tuple1)
p.append("Green")
Tuple1=tuple(p)
print ("Add Tuple element:", Tuple1)
p= list(Tuple1)
p.remove("Red")
Tuple1=tuple(p)
print("Remove Tuple element:",Tuple1)
Tuple2=("Purple","Neon","Sky Blue")
Tuple3=Tuple1+Tuple2
print("Combine Two tuples:", Tuple3)
```
