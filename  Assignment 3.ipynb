```python
def triangle():
    a=int(input("Enter first side:"))
    print(a)
    b=int(input("Enter second side:"))
    print(b) 
    c=int(input("Enter third side:"))
    print(c)
    if a^2+b^2==c^2 :
        print("The triangle is a right-angled triangle")
    else:
        print("The triangle is not a right-angled triangle")

triangle()   
```
